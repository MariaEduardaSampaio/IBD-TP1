```python
import pandas as pd
import csv
from io import StringIO

with open("filtered_indicators.csv", "r", encoding="utf-8-sig") as arquivo:
    linhas = [linha.strip() for linha in arquivo if linha.strip()]

# Remove as aspas que envolvem cada linha inteira
linhas_corrigidas = [
    linha[1:-1] if linha.startswith('"') and linha.endswith('"') else linha
    for linha in linhas
]

conteudo = StringIO("\n".join(linhas_corrigidas))

df = pd.read_csv(
    conteudo,
    sep=",",
    quotechar='"',
    doublequote=True
)

print(df.columns.tolist())
display(df.head())
```

```python

import csv
import pandas as pd
import sqlite3

# Ler e corrigir o CSV
with open("filtered_indicators.csv", "r", encoding="utf-8-sig", newline="") as arquivo:
    linhas = arquivo.readlines()

linhas_corrigidas = []

for linha in linhas:
    linha = linha.strip()

    if not linha:
        continue

    # Remove a primeira e a última aspa que envolvem a linha
    if linha.startswith('"') and linha.endswith('"'):
        linha = linha[1:-1]

    # Converte as aspas duplicadas em aspas normais
    linha = linha.replace('""', '"')

    linhas_corrigidas.append(linha)

# Salvar uma versão corrigida temporária
with open("filtered_indicators_corrigido.csv", "w", encoding="utf-8", newline="") as arquivo:
    arquivo.write("\n".join(linhas_corrigidas))

# Ler o CSV corrigido
df = pd.read_csv("filtered_indicators_corrigido.csv", sep=",", quotechar='"')

print("Colunas:", df.columns.tolist())
print("Dimensões:", df.shape)
display(df.head())
```

```python
df.to_sql(
    "filtered_indicators",
    con,
    if_exists="replace",
    index=False
)

con.commit()

display(pd.read_sql_query(
    "SELECT * FROM filtered_indicators",
    con
))
```
