```python

cur.execute("""
    CREATE TABLE Medicao (
        measurement_id INTEGER PRIMARY KEY AUTOINCREMENT,

        SurveyId TEXT NOT NULL,

        country_code TEXT NOT NULL,
        region_num_harmonized INTEGER NOT NULL,

        indicator_code TEXT NOT NULL,

        value REAL,
        standard_error REAL,

        FOREIGN KEY (SurveyId)
            REFERENCES Pesquisa(SurveyId),

        FOREIGN KEY (country_code, region_num_harmonized)
            REFERENCES "Região"(country_code, region_num_harmonized),

        FOREIGN KEY (indicator_code)
            REFERENCES Indicador(indicator_code),

        UNIQUE (
            SurveyId,
            country_code,
            region_num_harmonized,
            indicator_code
        )
    )
""")

con.commit()
print("Tabela Medicao criada com sucesso!")
```

```python
cur.execute("SELECT indicator_code FROM Indicador")
codigos = [linha[0] for linha in cur.fetchall()]

cur.execute('PRAGMA table_info("filtered_livwell")')
colunas = {linha[1] for linha in cur.fetchall()}

inseridos = 0

try:
    for codigo in codigos:
        coluna_valor = codigo.replace("-", ".")

        if coluna_valor not in colunas:
            raise ValueError(
                f"Coluna de valor não encontrada para {codigo}: "
                f"{coluna_valor}"
            )

        valor_sql = '"' + coluna_valor.replace('"', '""') + '"'

        coluna_se = coluna_valor + "_se"

        if coluna_se in colunas:
            se_sql = '"' + coluna_se.replace('"', '""') + '"'
        else:
            se_sql = "NULL"

        sql = f"""
            INSERT INTO Medicao (
                SurveyId,
                country_code,
                region_num_harmonized,
                indicator_code,
                value,
                standard_error
            )
            SELECT
                SurveyId,
                country_code,
                region_num_harmonized,
                ?,
                {valor_sql},
                {se_sql}
            FROM filtered_livwell
            WHERE {valor_sql} IS NOT NULL
        """

        cur.execute(sql, (codigo,))
        inseridos += cur.rowcount

    con.commit()
    print("Inserção concluída!")
    print("Medições inseridas:", inseridos)

except Exception:
    con.rollback()
    raise
```

```python


def qident(nome):
    """Protege nomes de colunas e tabelas no SQL SQLite."""
    return '"' + nome.replace('"', '""') + '"'

cur.execute('PRAGMA table_info("filtered_livwell")')
colunas_origem = {linha[1] for linha in cur.fetchall()}

cur.execute("SELECT indicator_code FROM Indicador")
codigos = [linha[0] for linha in cur.fetchall()]

total_esperado = 0
total_armazenado = 0
problemas = []

for codigo in codigos:
    if codigo in colunas_origem:
        coluna_valor = codigo
    else:
        coluna_valor = codigo.replace("-", ".")

    if coluna_valor not in colunas_origem:
        problemas.append((codigo, "Coluna de valor não encontrada"))
        continue

    valor_sql = qident(coluna_valor)
    coluna_se = coluna_valor + "_se"
    tem_se = coluna_se in colunas_origem

    se_sql = qident(coluna_se) if tem_se else "NULL"

    cur.execute(f"""
        SELECT COUNT(*)
        FROM filtered_livwell
        WHERE {valor_sql} IS NOT NULL
    """)
    esperado = cur.fetchone()[0]

    cur.execute("""
        SELECT COUNT(*)
        FROM Medicao
        WHERE indicator_code = ?
    """, (codigo,))
    armazenado = cur.fetchone()[0]

    total_esperado += esperado
    total_armazenado += armazenado

    cur.execute(f"""
        SELECT
            SUM(CASE
                WHEN m.value IS NOT s.{valor_sql} THEN 1
                ELSE 0
            END),
            SUM(CASE
                WHEN m.standard_error IS NOT {se_sql} THEN 1
                ELSE 0
            END)
        FROM Medicao m
        JOIN filtered_livwell s
          ON s.SurveyId = m.SurveyId
         AND s.country_code = m.country_code
         AND s.region_num_harmonized = m.region_num_harmonized
        WHERE m.indicator_code = ?
    """, (codigo,))

    divergencias_valor, divergencias_se = cur.fetchone()
    divergencias_valor = divergencias_valor or 0
    divergencias_se = divergencias_se or 0

    if esperado != armazenado:
        problemas.append((
            codigo,
            f"Quantidade: origem={esperado}, Medicao={armazenado}"
        ))

    if divergencias_valor:
        problemas.append((
            codigo, f"Valores divergentes: {divergencias_valor}"
        ))

    if divergencias_se:
        problemas.append((
            codigo, f"Erros padrão divergentes: {divergencias_se}"
        ))

print("Total esperado:", total_esperado)
print("Total armazenado:", total_armazenado)
print("Diferença:", total_esperado - total_armazenado)
print("Quantidade de problemas:", len(problemas))

for problema in problemas:
    print(problema)
```

```python
cur.execute("""
    SELECT SurveyId
    FROM filtered_livwell
    GROUP BY SurveyId
    HAVING COUNT(DISTINCT year) > 1
""")
print("Pesquisas com mais de um ano:", len(cur.fetchall()))

cur.execute("""
    SELECT COUNT(*)
    FROM filtered_livwell
    WHERE interview_year_mean IS NOT NULL
      AND ABS(interview_year_mean - year) > 0.000001
""")
print("Registros em que interview_year_mean difere de year:",
      cur.fetchone()[0])

cur.execute("""
    SELECT
        p.SurveyId,
        p.CMC_interview_mean,
        AVG(s.CMC_interview_mean) AS media_origem
    FROM Pesquisa p
    JOIN filtered_livwell s ON s.SurveyId = p.SurveyId
    GROUP BY p.SurveyId, p.CMC_interview_mean
""")

divergencias_cmc = []
for survey, cmc_tabela, cmc_origem in cur.fetchall():
    if cmc_origem is None:
        if cmc_tabela is not None:
            divergencias_cmc.append(survey)
    elif (
        cmc_tabela is None
        or abs(cmc_tabela - cmc_origem) > 0.000001
    ):
        divergencias_cmc.append(survey)

print("Pesquisas com média CMC divergente:", len(divergencias_cmc))
if divergencias_cmc:
    print(divergencias_cmc[:10])

cur.execute("""
    SELECT
        SUM(CASE WHEN value IS NULL THEN 1 ELSE 0 END),
        SUM(CASE WHEN standard_error IS NULL THEN 1 ELSE 0 END),
        COUNT(*)
    FROM Medicao
""")

nulos_valor, nulos_se, total = cur.fetchone()
print("Medições sem valor:", nulos_valor or 0)
print("Medições sem erro padrão:", nulos_se or 0)
print("Total de medições:", total)
```
