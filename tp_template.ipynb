# 3. Descrição dos dados (qual a URL? qual o domínio? como os dados foram processados?)

**URL** 
https://www.kaggle.com/datasets/konradb/wellbeing-of-women-in-52-countries

**Domínio** 
Saúde e condições de vida das mulheres, com foco em saúde reprodutiva e planejamento familiar e sua relação com características demográficas, escolaridade, autonomia nas decisões, violência doméstica, saúde materna, condições socioeconômicas, trabalho, habitação, nutrição e acesso à energia e informação. Os indicadores permitem comparar países, regiões e períodos de pesquisas DHS.

**Processamento**

O conjunto de dados utilizado foi o LivWell, disponibilizado na plataforma Kaggle, composto originalmente por 1.832 registros e 409 colunas.

Inicialmente, foram selecionados manualmente 100 indicadores relacionados ao domínio do trabalho, distribuídos em 16 categorias temáticas. A seleção foi registrada no arquivo `filtered_indicators.csv`, contendo os códigos, descrições e categorias dos indicadores.

Em seguida, os códigos selecionados foram associados às respectivas colunas do arquivo `livwell.csv`. Foram removidas as colunas de indicadores não selecionados, preservando os atributos de identificação de país, região, pesquisa e os erros-padrão disponíveis para os 100 indicadores escolhidos, sendo essa seleção registrada no arquivo `filtered_livwell.csv`. Nenhum registro da base original foi excluído nessa etapa.

Posteriormente, os dados foram reorganizados em quatro relações: Região, Pesquisa, Indicador e Medição. A transformação converteu as colunas de indicadores em registros individuais de medição, associando cada valor a uma região, pesquisa e indicador. Foram removidas apenas as medições sem valor observado, mantendo valores ausentes de erro-padrão como nulos.

Por fim, as relações foram inseridas em um banco SQLite, com definição de chaves primárias, chaves estrangeiras e restrições de unicidade. A integridade referencial foi verificada após a inserção dos dados.
